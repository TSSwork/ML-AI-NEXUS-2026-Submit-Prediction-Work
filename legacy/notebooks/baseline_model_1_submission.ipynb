```python
import pandas as pd
import numpy as np
import os

# 1. Load Data
base_path = '../../data/'
train = pd.read_csv(base_path + 'train.csv')
test = pd.read_csv(base_path + 'test.csv')
sub = pd.read_csv(base_path + 'sample_submission.csv')

# 2. SUBMISSION #1 — constant prevalence baseline
p0 = train["readmitted_30d"].mean()  # Should be exactly 0.125857
sub["readmitted_30d"] = p0

# 3. Format guards before submitting (Safety Checks)
assert (sub.patient_id.values == test.patient_id.values).all(), "❌ ID order mismatch between test and submission!"
assert ((sub.readmitted_30d > 0) & (sub.readmitted_30d < 1)).all(), "❌ Probabilities must be strictly between 0 and 1!"

# 4. Save to Kaggle Output Directory (/kaggle/working/)
output_dir = '../submissions/'  # Change this to '/kaggle/working/' if running on Kaggle
output_path = os.path.join(output_dir, 'submission_01_constant.csv')
sub.to_csv(output_path, index=False)

# 5. Print confirmation
print(sub.head())
print(f"\nconstant p = {p0:.6f}, rows = {len(sub)}")
print(f"✅ Saved to {output_path}")
print("👉 Go to the right-side panel in Kaggle -> 'Output' -> Download this file and submit it to the competition!")
```
